```python
import os
import pymongo

# Connection details: usa 'mongodb' dentro la rete Docker, o 'localhost' dall'esterno
mongo_host = os.getenv('MONGO_HOST', 'mongodb')
mongo_port = int(os.getenv('MONGO_PORT', 27017))
mongo_user = os.getenv('MONGO_USER', 'root')
mongo_password = os.getenv('MONGO_PASSWORD', 'example')

# Connect to MongoDB
try:
    client = pymongo.MongoClient(host=mongo_host, port=mongo_port, username=mongo_user, password=mongo_password, serverSelectionTimeoutMS=2000)
    client.server_info()
except Exception:
    # Fallback su localhost se eseguito direttamente sulla macchina host
    mongo_host = 'localhost'
    client = pymongo.MongoClient(host=mongo_host, port=mongo_port, username=mongo_user, password=mongo_password)

print(f"Connesso a MongoDB sull'host: {mongo_host}:{mongo_port}")

# Create a new database and collection
db = client['testdb']
collection = db['testcollection']

# Insert a document
document = {'name': 'test_document', 'value': 42}
inserted_id = collection.insert_one(document).inserted_id
print(f'Inserted document with id: {inserted_id}')

# Read the document back
retrieved_document = collection.find_one({'_id': inserted_id})
print(f'Retrieved document: {retrieved_document}')
```
